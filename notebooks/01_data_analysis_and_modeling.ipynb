## Problem Definition

### Objective

Predict a student's final academic performance using information available before the final grade is known.

### Target Variable

- Target: `G3`
- `G3` represents the student's final grade.

### Prediction Type

This project will initially treat the problem as a **regression problem**, because `G3` is a numerical grade from 0 to 20.

### Important Feature Decision

`G1` and `G2` will not be automatically removed or included.

We first need to define the exact prediction point:

- If prediction happens after the second period, `G1` and `G2` can be valid features.
- If prediction happens earlier, using `G1` or `G2` may not be appropriate.

The prediction point will be finalized before model training.

### Intended Use

The system is intended to provide an estimate of academic performance that could help identify students who may need additional academic support.

The prediction should be treated as decision-support information, not as a definitive judgment about a student.

## Prediction Point

The system predicts the student's final grade (`G3`) after the second-period grade (`G2`) is available.

- `G2` can be used as an input feature.
- `G3` is the target and must not be used as an input feature.

This defines a clear point in time for making the prediction and helps prevent target leakage.

## Desing the data preprocessing strategy

```python
numeric_features = df.drop(columns=["G3"]).select_dtypes(include="number").columns.tolist()
categorical_features = df.drop(columns=["G3"]).select_dtypes(include="str").columns.tolist()

print("Numeric features:")
print(numeric_features)

print("\nNumber of numeric features:", len(numeric_features))

print("\nCategorical features:")
print(categorical_features)

print("\nNumber of categorical features:", len(categorical_features))
```

```python
X = df.drop(columns=["G3"])
y = df["G3"]

print("X shape:", X.shape)
print("y shape:", y.shape)
print("G3 in X:", "G3" in X.columns)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)

print(preprocessor)
```

```python
preprocessor.fit(X_train)

print("Preprocessor fitted successfully.")
```

### Transform training and test data

```python
X_train_processed = preprocessor.transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("X_train_processed shape:", X_train_processed.shape)
print("X_test_processed shape:", X_test_processed.shape)
```

```python
import numpy as np

numeric_processed = X_train_processed[:, :len(numeric_features)]

print("Mean of first 5 numeric features:")
print(np.asarray(numeric_processed.mean(axis=0)).flatten()[:5])

print("\nStandard deviation of first 5 numeric features:")
print(np.asarray(numeric_processed.std(axis=0)).flatten()[:5])
```

### Baseline model

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

linear_regression_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)

print(linear_regression_pipeline)
```

```python
linear_regression_pipeline.fit(X_train, y_train)

print("Linear Regression model trained successfully.")
```

```python
y_pred = linear_regression_pipeline.predict(X_test)

print("First 10 actual grades:")
print(y_test.iloc[:10].values)

print("\nFirst 10 predicted grades:")
print(y_pred[:10])
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print(f"MAE:  {mae:.3f}")
print(f"RMSE: {rmse:.3f}")
print(f"R²:   {r2:.3f}")
```

### Cross-validation

```python
from sklearn.model_selection import KFold, cross_validate
```

```python
cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
models = {
    "Linear Regression": linear_regression_pipeline,
    "Decision Tree": decision_tree_pipeline,
    "Random Forest": random_forest_pipeline,
    "Gradient Boosting": gradient_boosting_pipeline
}

cv_results = {}

for name, model in models.items():
    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring={
            "mae": "neg_mean_absolute_error",
            "rmse": "neg_root_mean_squared_error",
            "r2": "r2"
        }
    )

    cv_results[name] = {
        "MAE": -scores["test_mae"].mean(),
        "RMSE": -scores["test_rmse"].mean(),
        "R²": scores["test_r2"].mean()
    }

cv_results
```

```python
cv_results_df = pd.DataFrame(cv_results).T

cv_results_df = cv_results_df.round(3)

cv_results_df
```

## Error Analysis

The residual plot was used to examine how the Gradient Boosting model's prediction errors are distributed.

Most predictions in the central range of approximately 8–16 have residuals relatively close to zero, indicating that the model generally predicts these observations with smaller errors.

Several larger negative residuals occur for predictions around 5–8. These represent cases where the actual G3 value is substantially lower than the model's prediction. Some of these observations have G3 = 0. The dataset alone does not establish why these students received a zero, so no specific cause is assumed.

There are also a small number of large positive residuals at low predicted values. These represent cases where the model substantially under-predicted the actual final grade.

The central portion of the residual plot does not show an obvious strong curved pattern. However, the presence of larger errors at some lower predicted values indicates that the model does not perform equally well for every type of observation.

Because the dataset contains only 395 observations, individual extreme residuals can have a noticeable effect on the evaluation. Further investigation of the largest errors is therefore useful before making changes to the modeling approach.

##### Next step: inspect the 10 largest prediction errors. This will tell us which actual student records the model is getting wrong instead of guessing why.

### Error Analysis Findings

The largest prediction errors were examined to determine whether they shared an obvious feature pattern.

Several observations with `G3 = 0` were substantially over-predicted. However, other observations with non-zero final grades were also substantially under-predicted. Therefore, the large errors cannot be attributed solely to the presence of zero-valued target observations.

The largest-error observations did not show one obvious common pattern across age, study time, failures, absences, previous grades, or the selected categorical features.

This suggests that some individual outcomes are difficult to predict from the available features. In particular, previous grades (`G1` and `G2`) provide strong overall predictive information, but they do not completely determine the final grade for every student.

No observations were removed based solely on their prediction error, and no causal explanation was assigned to the `G3 = 0` observations without supporting evidence.
